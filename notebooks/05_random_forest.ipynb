# Random Forest Classifier

## Goal
Combine many decision trees trained on varied samples/features. Voting across trees usually improves stability over one tree.

## Dataset
Breast cancer diagnostic dataset bundled with scikit-learn.

## How to run
Install the packages in `requirements.txt`, then run every cell from top to bottom. The fixed random seed makes the split and model repeatable.

## Steps
1. Load the example data.
2. Split it into training and test sets (or fit clusters when labels are unavailable).
3. Train the estimator on the training data.
4. Evaluate on held-out data or inspect cluster quality.

The test set is kept separate so the reported score estimates performance on unseen examples.

## 1. Import libraries and prepare the data

```python
from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, classification_report

data = load_breast_cancer(as_frame=True)
X, y = data.data, data.target
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, stratify=y, random_state=42)
model = RandomForestClassifier(n_estimators=200, random_state=42)
model.fit(X_train, y_train)
predictions = model.predict(X_test)
print(f"Accuracy: {accuracy_score(y_test, predictions):.3f}")
print(classification_report(y_test, predictions, target_names=data.target_names))
import pandas as pd
print(pd.Series(model.feature_importances_, index=data.feature_names).sort_values(ascending=False).head(10))
```

## 2. Interpret the result
The printed metric summarizes performance on data not used for fitting. For classification, read precision and recall for each class as well as accuracy. For clustering, a higher silhouette score generally indicates better-separated groups; inertia always decreases as k increases, so it should not be used alone to choose k. Results depend on the dataset and settings, and are educational examples rather than production models.
